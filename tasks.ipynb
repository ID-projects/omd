Есть список заказов:
- `status` -  статус заказа
  - `delivered` доставлен покупателю
  - `returned` возврат
- `amount` — сумма заказа.

Нужно найти:

- на какую сумму оформили возвраты
- кто хотя бы раз вернул заказ
- сколько заказов доставлено покупателю
- средний чек доставленных заказов

```python
orders = [
    {"id": 1, "buyer": "anya", "status": "delivered", "amount": 900},
    {"id": 2, "buyer": "boris", "status": "returned", "amount": 4_500},
    {"id": 3, "buyer": "anya", "status": "delivered", "amount": 1_500},
    {"id": 4, "buyer": "vera", "status": "delivered", "amount": 3_200},
    {"id": 5, "buyer": "boris", "status": "delivered", "amount": 700},
    {"id": 6, "buyer": "gleb", "status": "returned", "amount": 2_100},
]
```

```python
returned_sum= sum(order["amount"] for order in orders if order["status"] == "returned")
print(returned_sum)

returners_list = [order['buyer'] for order in orders
                    if order["status"] == "returned"]
print(returners_list)
returned_count = len(returners_list)
print(returned_count)

delivered_count = sum(1 for order in orders if order["status"] == "delivered")
print(delivered_count)
delivered_sum = sum(order["amount"] for order in orders if order["status"] == "delivered")
print(delivered_sum)
avg_delivered_sum = delivered_sum / delivered_count
print(avg_delivered_sum)
```

Есть список отзывов на товары. На каждой строке — по одному. Данные слегка битые, и названия товара записаны в разном регистре. `id` при этом один и тот же для одного и того же товара. Перед подсчётом приведите названия к одному формату, иначе они будут считаться как разные`.

Нужно найти:

- среднюю оценку каждого товара
- худший товар по средней оценке среди тех, у кого хотя бы два отзыва
- сколько отзывов на 1 или 2 звезды
- какую долю всех отзывов они составляют

```python
from collections import defaultdict

reviews = [
    {"id": 1, "product": "Чехол", "stars": 5},
    {"id": 1, "product": "Чехол", "stars": 3},
    {"id": 1, "product": "Чехол", "stars": 4},
    {"id": 2, "product": "Наушники", "stars": 2},
    {"id": 2, "product": "наушники", "stars": 2},
    {"id": 2, "product": "НАУШНИКИ", "stars": 5},
    {"id": 3, "product": "Планшет", "stars": 5},
    {"id": 4, "product": "Колонка", "stars": 4},
    {"id": 4, "product": "Колонка", "stars": 4},
    {"id": 5, "product": "Кабель", "stars": 1},
]

reviews_corrected = [{**item, "product": item["product"].lower()} for item in reviews]
# print(reviews_corrected)
avg_mark_by_product = {
    item["product"]: sum(
        review["stars"]
        for review in reviews_corrected
        if review["product"] == item["product"]
    )
    / len(
        [review for review in reviews_corrected if review["product"] == item["product"]]
    )
    for item in reviews_corrected
}

groups = defaultdict(lambda: {"count": 0, "stars_sum": 0})

for review in reviews_corrected:
    product = review["product"] 
    groups[product]["count"] += 1
    groups[product]["stars_sum"] += review["stars"]

result = {
    product: {
        "count": data["count"],
        "average": data["stars_sum"] / data["count"],
    }
    for product, data in groups.items()
    if data["count"] > 1
}

product_with_min_average= min(result, key=lambda x: result[x]["average"])
reviews_number = len(reviews_corrected)
reviews_1_2_count = sum(1 for review in reviews_corrected if review["stars"] in [1, 2])
bad_reviews_share = reviews_1_2_count / reviews_number
print(result)
print(product_with_min_average)
print(reviews_number)
print(reviews_1_2_count)
print(bad_reviews_share)
# print(avg_mark_by_product)
```
